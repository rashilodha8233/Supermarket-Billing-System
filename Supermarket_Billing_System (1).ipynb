```python
# ==============================================================================
# PROJECT: Super Market Billing System
# PROGRAMMING LANGUAGE: Python 3
# ==============================================================================

def supermarket_billing_system():
    # Welcome banner
    print("=" * 60)
    print("          WELCOME TO SUPERMARKET BILLING SYSTEM          ")
    print("=" * 60)

    # Requirement 1 & 5: Empty list to store multiple item dictionaries
    cart = []

    # Requirement 5: Continuous loop for real-time input
    while True:
        item_name = input("\nEnter Item Name (or type 'done' to finish): ").strip()

        # Stop condition for loop
        if item_name.lower() == 'done':
            break

        # Error handling for numeric inputs
        try:
            quantity = int(input(f"Enter quantity for '{item_name}': "))
            unit_price = float(input(f"Enter price per unit for '{item_name}' (₹): "))

            # Validation check
            if quantity <= 0 or unit_price <= 0:
                print("⚠️ Quantity and price must be greater than zero.")
                continue

            item_total = quantity * unit_price

            # Store item data as a dictionary entry
            cart.append({
                'name': item_name,
                'quantity': quantity,
                'unit_price': unit_price,
                'item_total': item_total
            })
            print(f"✅ Added: {quantity} x {item_name} @ ₹{unit_price:.2f} = ₹{item_total:.2f}")

        except ValueError:
            print("❌ Invalid input! Please enter numerical values for quantity and price.")

    # Edge case: If cart is empty
    if not cart:
        print("\n🛒 No items added. Billing system exited.")
        return

    # Requirement 3: Subtotal Calculation
    subtotal = sum(item['item_total'] for item in cart)

    # Requirement 2: Conditional Statements for Discounts
    if subtotal >= 2000:
        discount_percent = 15
    elif subtotal >= 1000:
        discount_percent = 10
    elif subtotal >= 500:
        discount_percent = 5
    else:
        discount_percent = 0

    discount_amount = (subtotal * discount_percent) / 100
    final_amount = subtotal - discount_amount

    # Requirement 4: Itemized Receipt Printing
    print("\n" + "=" * 60)
    print("                    OFFICIAL RECEIPT                     ")
    print("=" * 60)
    print(f"{'Item Name':<22} | {'Qty':<6} | {'Price(₹)':<10} | {'Total(₹)':<10}")
    print("-" * 60)

    for item in cart:
        print(f"{item['name']:<22} | {item['quantity']:<6} | {item['unit_price']:<10.2f} | {item['item_total']:<10.2f}")

    print("-" * 60)
    print(f"{'Subtotal:':<43} ₹{subtotal:.2f}")
    print(f"{'Discount (' + str(discount_percent) + '%):':<43} -₹{discount_amount:.2f}")
    print("=" * 60)
    print(f"{'FINAL AMOUNT DUE:':<43} ₹{final_amount:.2f}")
    print("=" * 60)
    print("          Thank you for shopping with us!               \n")

# Run the program
supermarket_billing_system()
```
